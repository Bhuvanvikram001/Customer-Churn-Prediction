## Day 10 - Model Evaluation

The tuned Logistic Regression model achieved an overall accuracy of 79.28% and a test ROC-AUC of 0.8415.

For the churn class (Class 1), the model achieved a precision of 64%, recall of 51%, and F1-score of 57%.

The recall of 51% means that the model correctly identifies approximately half of the customers who actually churn.

The precision of 64% indicates that among the customers predicted as churners, approximately 64% are actual churners.

The ROC-AUC score of 0.8415 indicates that the model has good ability to distinguish between customers who churn and those who do not.

From a business perspective, improving churn recall would be important because missing a customer who is likely to churn could result in a lost retention opportunity.

```python
y_prob_final[:10]
```

```python
import numpy as np
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.30, 0.35, 0.40, 0.45, 0.50, 0.55, 0.60]

threshold_results = []

for threshold in thresholds:
    
    y_pred_threshold = (
        y_prob_final >= threshold
    ).astype(int)
    
    precision = precision_score(
        y_test,
        y_pred_threshold,
        zero_division=0
    )
    
    recall = recall_score(
        y_test,
        y_pred_threshold,
        zero_division=0
    )
    
    f1 = f1_score(
        y_test,
        y_pred_threshold,
        zero_division=0
    )
    
    threshold_results.append([
        threshold,
        precision,
        recall,
        f1
    ])
```

```python
threshold_df = pd.DataFrame(
    threshold_results,
    columns=[
        "Threshold",
        "Precision",
        "Recall",
        "F1-Score"
    ]
)

threshold_df
```

```python
best_threshold_row = threshold_df.loc[
    threshold_df["F1-Score"].idxmax()
]

best_threshold_row
```

```python
best_threshold = best_threshold_row["Threshold"]

print("Best Threshold:", best_threshold)
print("Best F1-Score:", round(
    best_threshold_row["F1-Score"], 4
))
```

```python
y_pred_best = (
    y_prob_final >= best_threshold
).astype(int)
```

```python
final_precision = precision_score(
    y_test,
    y_pred_best
)

final_recall = recall_score(
    y_test,
    y_pred_best
)

final_f1 = f1_score(
    y_test,
    y_pred_best
)

print("Precision:", round(final_precision, 4))
print("Recall:", round(final_recall, 4))
print("F1-Score:", round(final_f1, 4))
```

```python
print("Precision:", round(final_precision * 100, 2), "%")
print("Recall:", round(final_recall * 100, 2), "%")
print("F1-Score:", round(final_f1 * 100, 2), "%")
```

## Day 11 - Classification Threshold Analysis

The default classification threshold of 0.50 was analyzed to determine whether churn detection could be improved.

Several classification thresholds from 0.30 to 0.60 were evaluated using Precision, Recall, and F1-Score.

The threshold of 0.35 achieved the highest F1-Score of 62.62%.

At this threshold:
- Precision = 55.60%
- Recall = 71.66%
- F1-Score = 62.62%

The recall improved from 51.34% at the default 0.50 threshold to 71.66% at the 0.35 threshold.

This indicates that the lower threshold allows the model to identify more customers who are likely to churn, which can be useful for customer retention campaigns.

The threshold of 0.35 will be used as the selected classification threshold for the final churn prediction system.
